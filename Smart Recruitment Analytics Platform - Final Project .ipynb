**Define Skill Keywords List**

```python
SKILL_KEYWORDS = [
    # IT & Software
    'python', 'java', 'javascript', 'c++', 'c#', 'sql', 'r', 'html', 'css',
    'react', 'node', 'django', 'flask', 'tensorflow', 'keras', 'pytorch',
    'machine learning', 'deep learning', 'data analysis', 'data science',
    'nlp', 'computer vision', 'aws', 'azure', 'docker', 'kubernetes', 'git',
    'mongodb', 'postgresql', 'mysql', 'tableau', 'power bi', 'excel',
    # Finance & Accounting
    'accounting', 'financial analysis', 'budgeting', 'forecasting', 'auditing',
    'taxation', 'tally', 'quickbooks', 'sap', 'ifrs', 'gaap',
    # HR & Business
    'recruitment', 'talent acquisition', 'payroll', 'employee relations',
    'performance management', 'communication', 'leadership', 'project management',
    'agile', 'scrum', 'microsoft office', 'crm', 'salesforce',
    # Engineering
    'autocad', 'matlab', 'solidworks', 'project planning', 'quality control',
    'civil engineering', 'mechanical engineering', 'electrical engineering',
    # Healthcare
    'patient care', 'clinical', 'nursing', 'pharmacology', 'medical billing',
    # Marketing & Media
    'seo', 'social media', 'content creation', 'adobe photoshop', 'illustrator',
    'google analytics', 'marketing strategy', 'branding', 'copywriting',
    # General
    'teamwork', 'problem solving', 'time management', 'customer service',
    'sales', 'negotiation', 'training', 'microsoft word', 'powerpoint',
]

print('Skill keywords defined!')
print(f'Total keywords: {len(SKILL_KEYWORDS)}')
```

**Define Education & Experience Extraction Functions**

```python
import re

EDUCATION_KEYWORDS = {
    'phd': 4,         'doctorate': 4,
    'master': 3,      'msc': 3,        'mba': 3,    'postgraduate': 3,
    'bachelor': 2,    'bsc': 2,        'ba': 2,     'b.tech': 2,
    'degree': 2,      'undergraduate': 2,
    'diploma': 1,     'certificate': 1, 'hnd': 1,
    'high school': 0, 'secondary': 0,  'gcse': 0
}

def extract_skills(text):
    text_lower = text.lower()
    found = [skill for skill in SKILL_KEYWORDS if skill in text_lower]
    return list(set(found))

def extract_experience_years(text):
    patterns = [
        r'(\d+)\+?\s+years? of experience',
        r'(\d+)\+?\s+years? experience',
        r'experience of (\d+)\+?\s+years?',
        r'(\d+)\+?\s+years? in',
    ]
    for pattern in patterns:
        match = re.search(pattern, text.lower())
        if match:
            return int(match.group(1))
    return 0

def extract_education_level(text):
    text_lower = text.lower()
    level = 0
    for keyword, val in EDUCATION_KEYWORDS.items():
        if keyword in text_lower:
            level = max(level, val)
    return level

def extract_features_from_cv(text):
    skills     = extract_skills(text)
    exp_years  = extract_experience_years(text)
    edu_level  = extract_education_level(text)
    skill_count = len(skills)
    return {
        'skills':       skills,
        'skill_count':  skill_count,
        'experience_years': exp_years,
        'education_level':  edu_level
    }

print('All extraction functions ready!')
```

**Keyword Matching Score - Method 1 Baseline**

```python
JD_KEYWORDS = [
    'python', 'sql', 'machine learning', 'data analysis', 'tensorflow',
    'pytorch', 'aws', 'azure', 'tableau', 'power bi', 'nlp', 'deep learning'
]

def keyword_match_score(cv_text):
    text_lower = cv_text.lower()
    matched = sum(1 for kw in JD_KEYWORDS if kw in text_lower)
    return matched / len(JD_KEYWORDS)

df['keyword_score'] = df['Resume_str'].apply(keyword_match_score)

print('Keyword matching done! (Method 1)')
print()
print(f'Average keyword score : {df["keyword_score"].mean():.4f}')
print(f'Max keyword score     : {df["keyword_score"].max():.4f}')
print(f'Min keyword score     : {df["keyword_score"].min():.4f}')
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_score, recall_score, f1_score,
    accuracy_score, confusion_matrix, classification_report
)
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rich.console import Console
from rich.table import Table

console = Console()

print('TensorFlow version:', tf.__version__)
print('All ANN libraries imported!')
```

**Prepare Data Using Same Engineered Features**

```python
FEATURES = [
    'Experience (Years)',
    'Education_Label',
    'Skill_Count',
    'Has_Certification',
    'JobRole_Label'
]

TARGET = 'Decision_Label'

X_ann = df2[FEATURES].values
y_ann = df2[TARGET].values

print('Features shape:', X_ann.shape)
print('Target shape  :', y_ann.shape)
print()
print('Target distribution:')
unique, counts = np.unique(y_ann, return_counts=True)
for u, c in zip(unique, counts):
    print(f'  {"Hire" if u == 1 else "Reject"}: {c}')
```

**Scale Features for ANN**

```python
scaler_ann = StandardScaler()
X_ann_scaled = scaler_ann.fit_transform(X_ann)

print('Feature scaling done (StandardScaler)!')
print()
print('Mean after scaling (should be ~0):')
print(np.round(X_ann_scaled.mean(axis=0), 4))
print()
print('Standard deviation after scaling (should be ~1):')
print(np.round(X_ann_scaled.std(axis=0), 4))

# Save scaler for Flask API
import joblib
joblib.dump(scaler_ann,
    '/content/drive/MyDrive/Waruni_Project/ann_scaler.pkl')
print()
print('ANN scaler saved!')
```

**Train / Test Split**

```python
X_train_ann, X_test_ann, y_train_ann, y_test_ann = train_test_split(
    X_ann_scaled, y_ann,
    test_size    = 0.2,
    random_state = 42,
    stratify     = y_ann
)

print('Train/Test split done!')
print(f'Training samples : {len(X_train_ann)}')
print(f'Testing  samples : {len(X_test_ann)}')
print()
print('Train target distribution:')
unique, counts = np.unique(y_train_ann, return_counts=True)
for u, c in zip(unique, counts):
    print(f'  {"Hire" if u == 1 else "Reject"}: {c}')
```

**Build the ANN Architecture**

```python
tf.random.set_seed(42)
ann_model = keras.Sequential([
    # Input layer
    layers.Input(shape=(len(FEATURES),)),

    # Hidden layer 1-64 neurons
    layers.Dense(64, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.3),
# Hidden layer 2-32 neurons
    layers.Dense(32, activation='relu'),
    layers.BatchNormalization(),
    layers.Dropout(0.2),
# Hidden layer 3-16 neurons
    layers.Dense(16, activation='relu'),
# Output-sigmoid for binary Hire/Reject
    layers.Dense(1, activation='sigmoid')
])

ann_model.compile(
    optimizer = 'adam',
    loss      = 'binary_crossentropy',
    metrics   = ['accuracy']
)
ann_model.summary()
```

**Train the ANN Model**

```python
early_stop = EarlyStopping(
    monitor             = 'val_loss',
    patience            = 10,
    restore_best_weights= True,
    verbose             = 1
)

history = ann_model.fit(
    X_train_ann, y_train_ann,
    epochs           = 100,
    batch_size       = 32,
    validation_split = 0.2,
    callbacks        = [early_stop],
    verbose          = 1
)

print()
print('ANN training complete!')
print(f'Stopped at epoch : {len(history.history["loss"])}')
```

**Plot Training History**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Loss
axes[0].plot(history.history['loss'],
             label='Train Loss', color='steelblue')
axes[0].plot(history.history['val_loss'],
             label='Val Loss',   color='tomato')
axes[0].set_title('ANN - Training vs Validation Loss')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()

# Accuracy
axes[1].plot(history.history['accuracy'],
             label='Train Accuracy', color='steelblue')
axes[1].plot(history.history['val_accuracy'],
             label='Val Accuracy',   color='tomato')
axes[1].set_title('ANN - Training vs Validation Accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].legend()

plt.tight_layout()
plt.show()
```

**Evaluate the ANN Model**

```python
y_pred_prob_ann = ann_model.predict(X_test_ann).flatten()
y_pred_ann      = (y_pred_prob_ann >= 0.5).astype(int)

accuracy_ann  = accuracy_score(y_test_ann,  y_pred_ann)
precision_ann = precision_score(y_test_ann, y_pred_ann)
recall_ann    = recall_score(y_test_ann,    y_pred_ann)
f1_ann        = f1_score(y_test_ann,        y_pred_ann)

table = Table(show_header=True, header_style='bold magenta',
              title='ANN Model Evaluation Results')
table.add_column('Metric',  justify='center')
table.add_column('Score',   justify='center')

table.add_row('Accuracy',  f'{accuracy_ann:.4f}')
table.add_row('Precision', f'{precision_ann:.4f}')
table.add_row('Recall',    f'{recall_ann:.4f}')
table.add_row('F1 Score',  f'{f1_ann:.4f}')

console.print(table)
print()
print('Full Classification Report:')
print(classification_report(y_test_ann, y_pred_ann,
      target_names=['Reject', 'Hire']))
```

```python
!pip install aif360

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from rich.console import Console
from rich.table import Table

from aif360.datasets import BinaryLabelDataset
from aif360.metrics import BinaryLabelDatasetMetric, ClassificationMetric
from aif360.algorithms.preprocessing import Reweighing

console = Console()

print('IBM AIF360 installed and imported!')
```

 Prepare Dataset for Fairness Analysis

```python
# Use the full scaled Dataset 2
df_fairness = df2[FEATURES + [TARGET]].copy()

# Create protected attribute from data
# Junior = Experience (Years) <= median → 0
# Senior = Experience (Years) >  median → 1
median_exp = df2['Experience (Years)'].median()

df_fairness['Experience_Group'] = (
    df2['Experience (Years)'] > median_exp
).astype(int)

print('Fairness dataset prepared!')
print(f'Median Experience     : {median_exp} years')
print(f'Junior candidates (0) : {(df_fairness["Experience_Group"] == 0).sum()}')
print(f'Senior candidates (1) : {(df_fairness["Experience_Group"] == 1).sum()}')
print()
print('Protected attribute : Experience_Group')
print('  0 = Junior (at or below median experience)')
print('  1 = Senior (above median experience)')
```

Create AIF360 Dataset

```python
# Scale features for AIF360
from sklearn.preprocessing import MinMaxScaler

scaler_fair  = MinMaxScaler()
feature_cols = FEATURES
X_fair       = scaler_fair.fit_transform(df_fairness[feature_cols])
X_fair_df    = pd.DataFrame(X_fair, columns=feature_cols)

# Build full dataframe for AIF360
aif_df = X_fair_df.copy()
aif_df['Decision_Label']    = df_fairness[TARGET].values
aif_df['Experience_Group']  = df_fairness['Experience_Group'].values

# Create BinaryLabelDataset
dataset = BinaryLabelDataset(
    df                  = aif_df,
    label_names         = ['Decision_Label'],
    protected_attribute_names = ['Experience_Group'],
    favorable_label     = 1,    # Hire
    unfavorable_label   = 0     # Reject
)

print('AIF360 BinaryLabelDataset created!')
print(f'Total candidates     : {len(aif_df)}')
print(f'Features             : {feature_cols}')
print(f'Label                : Decision_Label')
print(f'Protected attribute  : Experience_Group')
```

Measure Bias Before Mitigation

```python
# Define privileged and unprivileged groups
privileged_groups   = [{'Experience_Group': 1}]   # Senior
unprivileged_groups = [{'Experience_Group': 0}]   # Junior

# Calculate fairness metrics
metric_before = BinaryLabelDatasetMetric(
    dataset,
    privileged_groups   = privileged_groups,
    unprivileged_groups = unprivileged_groups
)

di_before  = metric_before.disparate_impact()
spd_before = metric_before.statistical_parity_difference()

print('Fairness Metrics BEFORE Mitigation:')
print()

table = Table(show_header=True, header_style='bold magenta',
              title='Bias Metrics — Before Reweighing')
table.add_column('Metric',       justify='center')
table.add_column('Value',        justify='center')
table.add_column('Fair Range',   justify='center')
table.add_column('Status',       justify='center')

table.add_row(
    'Disparate Impact',
    f'{di_before:.4f}',
    '0.8 – 1.2',
    'Fair' if 0.8 <= di_before <= 1.2 else 'Biased'
)
table.add_row(
    'Statistical Parity Difference',
    f'{spd_before:.4f}',
    '-0.1 – 0.1',
    'Fair' if -0.1 <= spd_before <= 0.1 else 'Biased'
)

console.print(table)
print()
print('Disparate Impact explanation:')
print('  < 0.8 = Junior candidates unfairly disadvantaged')
print('  > 1.2 = Senior candidates unfairly disadvantaged')
print('  0.8–1.2 = Fair treatment')
```

Visualise Hire Rate Before Mitigation

```python
# Show hire rate for Junior vs Senior before mitigation

hire_rates_before = aif_df.groupby('Experience_Group')['Decision_Label'].mean()

plt.figure(figsize=(7, 4))
bars = plt.bar(
    ['Junior (Group 0)', 'Senior (Group 1)'],
    hire_rates_before.values,
    color = ['tomato', 'steelblue'],
    edgecolor = 'white'
)
plt.axhline(
    y     = hire_rates_before.mean(),
    color = 'black',
    linestyle = '--',
    label = 'Average hire rate'
)
plt.title('Hire Rate by Experience Group - Before Mitigation')
plt.ylabel('Hire Rate')
plt.ylim(0, 1)
plt.legend()

for bar, val in zip(bars, hire_rates_before.values):
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height() + 0.01,
             f'{val:.2%}', ha='center', fontsize=11)

plt.tight_layout()
plt.show()

print(f'Junior hire rate : {hire_rates_before[0]:.2%}')
print(f'Senior hire rate : {hire_rates_before[1]:.2%}')
print(f'Difference       : {abs(hire_rates_before[1] - hire_rates_before[0]):.2%}')
```

 Apply Reweighing Mitigation

```python
RW = Reweighing(
    unprivileged_groups = unprivileged_groups,
    privileged_groups   = privileged_groups
)

dataset_reweighed = RW.fit_transform(dataset)

print('Reweighing applied!')
print()
print('Reweighing assigns sample weights learned from data')
print('so the model treats Junior and Senior candidates more fairly.')
print()

# Show weight distribution
weights = dataset_reweighed.instance_weights
print(f'Min weight  : {weights.min():.4f}')
print(f'Max weight  : {weights.max():.4f}')
print(f'Mean weight : {weights.mean():.4f}')
```

Measure Bias After Mitigation

```python
metric_after = BinaryLabelDatasetMetric(
    dataset_reweighed,
    privileged_groups   = privileged_groups,
    unprivileged_groups = unprivileged_groups
)

di_after  = metric_after.disparate_impact()
spd_after = metric_after.statistical_parity_difference()

table = Table(show_header=True, header_style='bold magenta',
              title='Bias Metrics — After Reweighing')
table.add_column('Metric',      justify='center')
table.add_column('Value',       justify='center')
table.add_column('Fair Range',  justify='center')
table.add_column('Status',      justify='center')

table.add_row(
    'Disparate Impact',
    f'{di_after:.4f}',
    '0.8 – 1.2',
    'Fair' if 0.8 <= di_after <= 1.2 else 'Biased'
)
table.add_row(
    'Statistical Parity Difference',
    f'{spd_after:.4f}',
    '-0.1 – 0.1',
    'Fair' if -0.1 <= spd_after <= 0.1 else 'Biased'
)

console.print(table)
```

Before vs After Comparison Chart

```python
metrics   = ['Disparate Impact', 'Statistical Parity Difference']
before    = [di_before,  spd_before]
after     = [di_after,   spd_after]

x     = np.arange(len(metrics))
width = 0.35

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x - width/2, before, width, label='Before Reweighing', color='tomato')
ax.bar(x + width/2, after,  width, label='After Reweighing',  color='steelblue')

ax.set_title('Fairness Metrics - Before vs After Reweighing')
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.set_ylabel('Metric Value')
ax.legend()

# Fair range lines
ax.axhline(y=0.8,  color='green', linestyle='--', alpha=0.5, label='Fair lower bound')
ax.axhline(y=1.2,  color='green', linestyle='--', alpha=0.5, label='Fair upper bound')
ax.axhline(y=0.0,  color='gray',  linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()

print()
print('Disparate Impact closer to 1.0 = fairer model')
print('Statistical Parity Difference closer to 0.0 = fairer model')
```

Retrain XGBoost with Fair Weights

```python
from sklearn.model_selection import train_test_split, GridSearchCV
import xgboost as xgb

# AIF360 includes protected attribute in features — get correct column names
feature_cols_rw = dataset_reweighed.feature_names

X_rw       = pd.DataFrame(
    dataset_reweighed.features,
    columns = feature_cols_rw    # use AIF360's own column names
)
y_rw       = dataset_reweighed.labels.ravel()
weights_rw = dataset_reweighed.instance_weights

print('Reweighed data shape:', X_rw.shape)
print('Columns:', feature_cols_rw)
print()

# Split
X_train_rw, X_test_rw, y_train_rw, y_test_rw, w_train_rw, w_test_rw = \
    train_test_split(
        X_rw, y_rw, weights_rw,
        test_size    = 0.2,
        random_state = 42,
        stratify     = y_rw
    )

# GridSearchCV learns best params from reweighed data
param_grid = {
    'n_estimators'  : [50, 100, 200],
    'max_depth'     : [3, 4, 5],
    'learning_rate' : [0.05, 0.1, 0.2]
}

grid_rw = GridSearchCV(
    estimator  = xgb.XGBClassifier(
                     eval_metric  = 'logloss',
                     random_state = 42
                 ),
    param_grid = param_grid,
    cv         = 5,
    scoring    = 'f1',
    n_jobs     = -1,
    verbose    = 1
)

grid_rw.fit(X_train_rw, y_train_rw,
            sample_weight = w_train_rw)

xgb_fair = grid_rw.best_estimator_

print()
print('Fair XGBoost trained with reweighed data!')
print('Best params:', grid_rw.best_params_)
print('Best F1    :', round(grid_rw.best_score_, 4))
```
